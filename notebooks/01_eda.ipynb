```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('darkgrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

```python
manga_df = pd.read_csv('../data/raw/manga_metadata.csv')
sales_df = pd.read_csv('../data/raw/manga_sales.csv')
users_df = pd.read_csv('../data/raw/users.csv')
interactions_df = pd.read_csv('../data/raw/user_interactions.csv')

print(f"Manga: {len(manga_df)} titles")
print(f"Sales records: {len(sales_df)} rows")
print(f"Users: {len(users_df)} users")
print(f"Interactions: {len(interactions_df)} interactions")
```

Explore manga metadata

```python
manga_df.head()

genre_counts = manga_df['genre'].str.split(',', expand=True).stack().value_counts()
plt.figure(figsize=(12, 6))
genre_counts.head(15).plot(kind='barh')
plt.title('Top 15 Manga Genres')
plt.xlabel('Count')
plt.tight_layout()
plt.show()
```

Sales over time

```python
sales_df['date'] = pd.to_datetime(sales_df['date'])
monthly_sales = sales_df.groupby(sales_df['date'].dt.to_period('M'))['sales_count'].sum()

plt.figure(figsize=(14, 6))
monthly_sales.plot()
plt.title('Total Manga Sales Over Time')
plt.ylabel('Sales Count')
plt.xlabel('Date')
plt.tight_layout()
plt.show()
```

Top selling manga

```python
top_manga = sales_df.groupby('title')['sales_count'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(10, 6))
top_manga.plot(kind='barh')
plt.title('Top 10 Best Selling Manga (Total Sales)')
plt.xlabel('Total Sales')
plt.tight_layout()
plt.show()
```

User interaction patterns

```python
print(interactions_df['interaction_type'].value_counts())
print(f"\nAverage rating: {interactions_df['rating'].mean():.2f}")
```

Sales trajectory example

```python
sample_manga = sales_df[sales_df['manga_id'] == sales_df['manga_id'].iloc[0]]
plt.figure(figsize=(12, 6))
for volume in sample_manga['volume'].unique()[:5]:
    vol_data = sample_manga[sample_manga['volume'] == volume]
    plt.plot(vol_data['week_number'], vol_data['sales_count'], marker='o', label=f'Volume {volume}')
plt.title(f"Sales Trajectory: {sample_manga['title'].iloc[0]}")
plt.xlabel('Week Number')
plt.ylabel('Sales Count')
plt.legend()
plt.tight_layout()
plt.show()
```
